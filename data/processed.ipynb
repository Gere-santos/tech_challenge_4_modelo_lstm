```python
tz = pytz.timezone("America/Sao_Paulo")
def fetch_efinance_data():
    df = yf.download('BBAS3.SA', start= '2005-01-01', end = datetime.now(tz).strftime("%Y-%m-%d"))
    return df
df = fetch_efinance_data()


date_str = datetime.now(tz).strftime("%Y_%m_%d")
file_name = f"{date_str}.csv"
df.to_csv(fr"C:\Users\gerem\Desktop\projeto_tc_fiap_4\data\raw\data_{file_name}" )
```
